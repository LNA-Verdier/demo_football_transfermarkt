```python
import os
import sys
import logging
project_root = os.path.abspath(os.path.join(os.getcwd(), "../.."))
sys.path.append(project_root)
from config.conf import create_spark_session
from config.conf import config
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

```python
spark = create_spark_session()
```

```python
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
logger = logging.getLogger(__name__)
```

```python
csv_files = [
    "appearances.csv",
    "club_games.csv",
    "competitions.csv",
    "game_events.csv",
    "game_lineups.csv",
    "games.csv",
    "player_valuations.csv",
    "players.csv",
    "transfers.csv"
]

for file in csv_files:
    name = file.replace(".csv", "")
    input_path = f"{config.root_path}{config.landing_zone}{file}"
    output_path = f"{config.root_path}{config.bronze_zone}{name}"
    
    try:
        logger.info(f"Lecture de {file} depuis {input_path}")
        df = spark.read.option("header", "true").csv(input_path)

        logger.info(f"Écriture du fichier {name} au format Parquet vers {output_path}")
        df.write.mode("overwrite").parquet(output_path)

        logger.info(f"✅ Fichier {file} traité avec succès.\n")
    except Exception as e:
        logger.error(f"❌ Erreur lors du traitement du fichier {file} : {e}")
```
